```python
def add_target(team):
    team = team.copy()
    team.loc[:, 'target'] = team['won'].shift(-1)
    team.loc[:, 'target'] = team['target'].fillna(2)
    return team

df = df.groupby('team', group_keys=False).apply(add_target, include_groups=True)
```

```python
df['target'] = df['target'].astype(int, errors='ignore')
```

```python
df = df[df['target']!= 2].copy()
```

0 means loss 1 means win 2 means there is no data on the next game

```python
def shift_col(team, col_name):
    next_col = team[col_name].shift(-1)
    return next_col

def add_col(df, col_name):
    return df.groupby("team", group_keys=False).apply(lambda x: shift_col(x, col_name))

df["home_next"] = add_col(df, "home")
df["team_opp_next"] = add_col(df, "team_opp")
df["date_next"] = add_col(df, "date")

nulls = pd.isnull(df).sum()
nulls = nulls[nulls > 0]
print("\nNull Values:\n", nulls)

columns_to_keep = ['team_opp_next', 'date_next', 'team', 'season', 'date']
valid_columns = df.columns[~df.columns.isin(nulls.index) | df.columns.isin(columns_to_keep)]

df = df[valid_columns].copy()
```

```python
df
```

```python
df['date'] = pd.to_datetime(df['date'])

df['rest_days'] = df.groupby('team')['date'].diff().dt.days.fillna(3)

df['point_diff'] = df['pts'] - df['pts_opp']

df['win_streak'] = df.groupby('team')['won'].transform(lambda x: x.rolling(5, min_periods=1).sum())

df['avg_point_diff_10'] = df.groupby('team')['point_diff'].transform(
    lambda x: x.rolling(10, min_periods=1).mean()
)

opp_wins = df.groupby('team')['won'].transform(
    lambda x: x.rolling(10, min_periods=1).mean()
)

df['pace_diff'] = df['total'] - df['total_opp']

def calc_home_win_rate(group):
    home_games = group[group['home'] == 1].copy()
    home_games['home_win_rate_10'] = home_games['won'].rolling(10, min_periods=1).mean()
    group = group.copy()
    group.loc[group['home'] == 1, 'home_win_rate_10'] = home_games['home_win_rate_10'].values
    group['home_win_rate_10'] = group['home_win_rate_10'].fillna(0.5)
    return group

df = df.groupby('team', group_keys=False).apply(calc_home_win_rate, include_groups=True)
```

```python
removed_columns = ['season', 'date', 'won', 'target', 'team', 'team_opp',
                   'team_opp_next', 'date_next', 'home_next', 'next_opponent']
selected_columns = df.columns[~df.columns.isin(removed_columns)]
```

```python
numeric_columns = df[selected_columns].select_dtypes(include=['number']).columns.tolist()
selected_columns = numeric_columns
```

```python
print(df.columns.tolist())
```

```python
def rolling_av_features(team, windows=[3, 5, 10, 20]):

    team = team.copy()


    rolling_dfs = []

    for window in windows:
        rolling = team[selected_columns].rolling(window).mean()
        rolling.columns = [f'{col}_rolling_{window}' for col in rolling.columns]
        rolling_dfs.append(rolling)

    all_rolling = pd.concat(rolling_dfs, axis=1)

    team = pd.concat([team, all_rolling], axis=1)

    return team

df = df.groupby('team', group_keys=False).apply(
    lambda x: rolling_av_features(x, windows=[3, 5, 10, 20]),
    include_groups=True
)
```

```python
print("\n" + "="*70)
print("MERGING OPPONENT NEXT GAME FEATURES")
print("="*70)

rolling_cols = [col for col in df.columns if 'rolling' in col]
print(f"Number of rolling features: {len(rolling_cols)}")

# Check columns exist
print(f"'team' exists: {'team' in df.columns}")
print(f"'team_opp_next' exists: {'team_opp_next' in df.columns}")
print(f"'date_next' exists: {'date_next' in df.columns}")

df_copy = df.copy()

full = df.merge(
    df_copy[rolling_cols + ["team_opp_next", "date_next", "team"]],
    left_on=["team", "date_next"],
    right_on=["team_opp_next", "date_next"],
    suffixes=('', '_opp_next')
)

print(f"Rows before merge: {len(df)}")
print(f"Rows after merge: {len(full)}")

if len(full) >= len(df) * 0.8:
    df = full.copy()
    print("Merge successful!")
    df = df.drop(['team_opp_next'], axis=1, errors='ignore')
    if 'team_x' in df.columns:
        df = df.rename(columns={'team_x': 'team', 'team_y': 'next_opponent'})
else:
    print("Merge issue - continuing without opponent next features")

df = df.dropna().copy()
print(f"Final shape after dropna: {df.shape}")
```

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import TimeSeriesSplit
from sklearn.linear_model import RidgeClassifier
from sklearn.metrics import accuracy_score
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.ensemble import RandomForestClassifier

scaler = MinMaxScaler()
df[selected_columns] = scaler.fit_transform(df[selected_columns])

rr = RidgeClassifier(alpha=1)
split = TimeSeriesSplit(n_splits=3)

feature_cols = df.select_dtypes(include=['number']).columns.tolist()

cols_to_exclude = ['season', 'won', 'target']
feature_cols = [col for col in feature_cols if col not in cols_to_exclude]

rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(df[feature_cols], df['target'])

feature_importance = pd.DataFrame({
    'feature': feature_cols,
    'importance': rf.feature_importances_
}).sort_values('importance', ascending=False)

predictors = feature_importance.head(50)['feature'].tolist()

print(f"\nSelected {len(predictors)} features")
print("Top 10 most important features:")
print(feature_importance.head(10))
```

```python
print("Most Important Features Graph: ")

top_10 = feature_importance.head(10)

fig = go.Figure(go.Bar(
    x=top_10['feature'],
    y=top_10['importance'],
    marker=dict(
        color=top_10['importance'],
        colorscale='Viridis',
        showscale=True,
        colorbar=dict(title="Importance")
    ),
    text=[f"{v:.4f}" for v in top_10['importance']],
    textposition='outside'
))

fig.update_layout(
    title='Top 10 Most Important Features for NBA Game Prediction',
    xaxis_title='Feature',
    yaxis_title='Feature Importance',
    height=600,
    showlegend=False,
    xaxis_tickangle=-45
)
fig.show()
```

```python
def backtest(data, model, predictors, start=2, step=1):
    all_predictions = []

    seasons = sorted(data["season"].unique())

    for i in range(start, len(seasons), step):
        current_season = seasons[i]
        df_train = data[data["season"] < current_season]
        df_test = data[data["season"] == current_season]

        model.fit(df_train[predictors], df_train["target"])
        preds = model.predict(df_test[predictors])
        preds = pd.Series(preds, index=df_test.index)
        combined = pd.concat([df_test["target"], preds], axis=1)
        combined.columns = ["actual", "prediction"]

        all_predictions.append(combined)

    return pd.concat(all_predictions)
#
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import GridSearchCV

print("Tuning RidgeClassifier...")
split = TimeSeriesSplit(n_splits=3)
param_grid = {'alpha': [0.1, 0.5, 1.0, 2.0, 5.0]}
grid_search = GridSearchCV(
    RidgeClassifier(),
    param_grid,
    cv=split,
    scoring='accuracy',
    n_jobs=-1
)

grid_search.fit(df[predictors], df['target'])
best_rr = grid_search.best_estimator_
print(f"Best alpha: {grid_search.best_params_['alpha']}")

print("Training GradientBoostingClassifier...")
gb = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=5,
    random_state=42
)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

predictions_rr = backtest(df, best_rr, predictors)
accuracy_rr = accuracy_score(predictions_rr['actual'], predictions_rr['prediction'])

predictions_gb = backtest(df, gb, predictors)
accuracy_gb = accuracy_score(predictions_gb['actual'], predictions_gb['prediction'])

home_win_rate = df.groupby("home")["won"].mean()
baseline = home_win_rate.mean()

print("FINAL RESULTS: ")

print(f"\nRidgeClassifier Accuracy: {accuracy_rr:.4f}")
print(f"GradientBoosting Accuracy: {accuracy_gb:.4f}")
print(f"Baseline (home team win rate): {baseline:.4f}")
print(f"\nBest Model: {'RidgeClassifier' if accuracy_rr > accuracy_gb else 'GradientBoosting'}")
print(f"Best Accuracy: {max(accuracy_rr, accuracy_gb):.4f}")
print(f"Improvement over baseline: {(max(accuracy_rr, accuracy_gb) - baseline):.4f}")

predictions = predictions_gb if accuracy_gb > accuracy_rr else predictions_rr
accuracy = max(accuracy_gb, accuracy_rr)

print("\nPrediction distribution:")
print(predictions['prediction'].value_counts().sort_index())

print("\nActual distribution:")
print(predictions['actual'].value_counts().sort_index())

print("\nClassification Report:")
print(classification_report(predictions['actual'], predictions['prediction']))

print("\nConfusion Matrix:")
print(confusion_matrix(predictions['actual'], predictions['prediction']))
```
